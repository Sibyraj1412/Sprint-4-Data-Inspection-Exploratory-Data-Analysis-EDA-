## Dataset Documentation

**Dataset name:** Healthcare Dataset

**Dataset source:** Kaggle — [Healthcare Dataset by prasad22](https://www.kaggle.com/datasets/prasad22/healthcare-dataset)

**Business/domain:** Healthcare — patient admission records from a hospital system, covering demographics, medical conditions, admission/discharge details, billing, and test outcomes.

**Number of rows and columns:** see the code cell below for the exact counts.

**Meaning of important columns:**
- `Name` — patient's name (identifier-like, not useful for analysis).
- `Age` — patient's age in years (numerical).
- `Gender` — patient's gender (categorical).
- `Blood Type` — patient's blood group, e.g. A+, O- (categorical).
- `Medical Condition` — diagnosed condition, e.g. Diabetes, Cancer (categorical).
- `Date of Admission` — the date the patient was admitted (date/time).
- `Doctor` — attending doctor's name (identifier-like).
- `Hospital` — hospital/facility name (categorical/identifier-like).
- `Insurance Provider` — insurance company covering the patient (categorical).
- `Billing Amount` — amount billed for the stay (numerical, continuous).
- `Room Number` — assigned room number (numerical, but functions as an identifier rather than a measurable quantity).
- `Admission Type` — Emergency, Elective, or Urgent (categorical).
- `Discharge Date` — the date the patient was discharged (date/time).
- `Medication` — medication prescribed (categorical).
- `Test Results` — Normal, Abnormal, or Inconclusive (categorical).

**Target variable:** `Test Results` is used as the target variable in this project — predicting whether a patient's test outcome is Normal, Abnormal, or Inconclusive based on their other attributes. (`Medical Condition` is a reasonable alternative target.)

**Why this dataset was selected:** It combines a realistic mix of numerical, categorical, and date/time columns, has a large enough number of records for meaningful analysis, and includes real-world data-quality issues (such as inconsistent text casing in names and duplicate-like records) that make it suitable for practicing data inspection and cleaning.

## What is a Dataset?

### Concept
A dataset is an organized collection of data, typically structured as a table where each row is an observation and each column is an attribute of that observation.

### Example
This healthcare dataset is a table where each row represents one patient's hospital admission record.

### Implementation
See the code cell below.

```python
df.head()
```

### Interpretation
Each row is a single, self-contained observation (one patient's admission), and each column describes one property of that observation.

## What is a Feature?

### Concept
A feature (also called a variable or attribute) is any column in the dataset that describes a characteristic of an observation.

### Example
`Age`, `Gender`, and `Billing Amount` are all features of a patient's record.

### Implementation
See the code cell below.

```python
print("Features:", df.columns.tolist())
```

### Interpretation
Every column apart from the target (if one is used for prediction) can serve as an input feature for analysis or modeling.

## What is a Target?

### Concept
The target (or label) is the specific column a predictive model tries to estimate or classify, based on the other features.

### Example
In this dataset, `Test Results` is treated as the target — the outcome we would want to predict.

### Implementation
See the code cell below.

```python
print(df["Test Results"].value_counts())
```

### Interpretation
The target column has a small number of distinct classes (Normal, Abnormal, Inconclusive), which makes it suitable for a classification task.

## Independent Variables

### Concept
Independent variables are the input features used to predict or explain the target; their values are not influenced by the target.

### Example
`Age`, `Gender`, `Medical Condition`, `Admission Type`, and `Billing Amount` are independent variables with respect to `Test Results`.

### Implementation
See the code cell below.

```python
independent_vars = [col for col in df.columns if col != "Test Results"]
print("Independent Variables:", independent_vars)
```

### Interpretation
These are the columns that would be fed into a model to make a prediction about the target.

## Dependent Variables

### Concept
The dependent variable is the outcome that depends on (and is predicted from) the independent variables. It is another name for the target.

### Example
`Test Results` is the dependent variable in this dataset.

### Implementation
See the code cell below.

```python
dependent_var = "Test Results"
print("Dependent Variable:", dependent_var)
```

### Interpretation
Its value is assumed to change in response to the combination of independent variables describing each patient.

## Numerical Features

### Concept
Numerical features contain measurable quantities and support mathematical operations like averaging or comparison.

### Example
`Age` and `Billing Amount` are numerical features.

### Implementation
See the code cell below.

```python
numerical_features = df.select_dtypes(include="number").columns.tolist()
print("Numerical Features:", numerical_features)
```

### Interpretation
These columns can be summarized with statistics like mean, median, and standard deviation.

## Categorical Features

### Concept
Categorical features contain a limited set of distinct labels or categories rather than numeric quantities.

### Example
`Gender`, `Blood Type`, and `Insurance Provider` are categorical features.

### Implementation
See the code cell below.

```python
categorical_features = df.select_dtypes(include="object").columns.tolist()
print("Categorical Features:", categorical_features)
```

### Interpretation
These columns represent group membership rather than measurable quantities, and are best summarized with counts or proportions.

## Discrete Features

### Concept
Discrete features take on a countable, finite (or countably infinite) set of values — often whole numbers.

### Example
`Room Number` is discrete — rooms are assigned whole numbers with no meaningful values in between.

### Implementation
See the code cell below.

```python
print("Unique Room Numbers:", df["Room Number"].nunique())
print(df["Room Number"].unique()[:10])
```

### Interpretation
Room numbers are counted, not measured — there is no meaningful "half room," which is the hallmark of a discrete variable.

## Continuous Features

### Concept
Continuous features can take any value within a range, including fractional or decimal values.

### Example
`Billing Amount` is continuous — it can take any decimal value within a range.

### Implementation
See the code cell below.

```python
print(df["Billing Amount"].describe())
```

### Interpretation
Billing amounts vary smoothly across a wide numeric range, unlike the fixed, countable values seen in discrete features.

## Identifier Columns

### Concept
Identifier columns uniquely (or near-uniquely) label each record and are not useful as predictive features.

### Example
`Name` and `Doctor` act as identifiers rather than predictive features.

### Implementation
See the code cell below.

```python
print("Unique Names:", df["Name"].nunique(), "out of", len(df), "rows")
print("Unique Doctors:", df["Doctor"].nunique())
```

### Interpretation
A high ratio of unique values to total rows signals that a column is functioning as an identifier rather than a meaningful predictive feature.

## Target Variable

### Concept
The target variable is the specific column selected as the prediction goal for this project.

### Example
`Test Results` — classifying whether a patient's test outcome is Normal, Abnormal, or Inconclusive.

### Implementation
See the code cell below.

```python
df["Test Results"].value_counts(normalize=True)
```

### Interpretation
Viewing the proportions shows whether the classes are balanced or skewed, which matters for later modeling choices.

## Metadata

### Concept
Metadata is "data about data" — information describing the dataset itself, such as column names, data types, and source, rather than the observations.

### Example
The dataset documentation section above (name, source, domain, column meanings) is metadata.

### Implementation
See the code cell below.

```python
df.info()
```

### Interpretation
`info()` is a convenient way to view key metadata: column names, data types, and non-null counts, all in one place.
