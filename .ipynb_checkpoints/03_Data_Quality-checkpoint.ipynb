# Notebook 3 – Data Quality Assessment

This notebook investigates data-quality issues in the Healthcare dataset: missing values, duplicate records, and invalid values.

## Load the Dataset

```python
import pandas as pd
import numpy as np
import os

possible_names = ["healthcare_dataset.csv", "Healthcare_dataset.csv", "Healthcare Dataset.csv", "healthcare-dataset.csv"]
data_path = None

for name in possible_names:
    if os.path.exists(name):
        data_path = name
        break

if data_path is None:
    print("Could not find the dataset automatically.")
    print("Current folder:", os.getcwd())
    print("Files in this folder:", os.listdir())
    print("\nFix: set data_path manually below to the exact filename you see above,")
    print("e.g. data_path = 'my_actual_filename.csv'")
else:
    print("Found dataset at:", data_path)

# If the automatic search above didn't find it, uncomment and edit the line below:
# data_path = "PUT_YOUR_EXACT_FILENAME_HERE.csv"

df = pd.read_csv(data_path)
df.head()
```

# Part 1: Missing Values

## Number of Missing Values

### Concept
A missing value is an empty or null entry where data should exist. Counting them per column is the first step in assessing data quality.

### Implementation
See the code cell below.

```python
missing_counts = df.isnull().sum()
print("Missing Values per Column:")
print(missing_counts)
```

### Interpretation
Any column with a count greater than zero has incomplete records that may need imputation or removal before analysis.

## Percentage of Missing Values

### Concept
Expressing missing values as a percentage of total rows makes it easier to judge severity across columns of the same dataset.

### Implementation
See the code cell below.

```python
missing_percent = (df.isnull().sum() / len(df)) * 100
print("Percentage of Missing Values per Column:")
print(missing_percent.round(2))
```

### Interpretation
A column missing less than ~5% of its values is usually safe to impute, while columns missing a large share may need to be dropped or investigated for a systemic cause.

## Missing Values by Column (Summary Table)

### Concept
Combining the count and percentage into a single table gives a clear, sortable summary of where data quality issues are concentrated.

### Implementation
See the code cell below.

```python
missing_summary = pd.DataFrame({
    "Missing Count": missing_counts,
    "Missing Percentage": missing_percent.round(2)
})
missing_summary = missing_summary[missing_summary["Missing Count"] > 0].sort_values("Missing Count", ascending=False)
print(missing_summary)
```

### Interpretation
Sorting by missing count highlights the columns that need the most attention first. An empty table here means no column has any missing values by the strict `isnull()` definition — though null-like placeholder text can still hide inside "complete" columns (checked later in this notebook).

## Missing Value Patterns

### Concept
Missing values are sometimes not random — they can be concentrated in specific rows, correlated across columns, or linked to another feature (e.g., all missing `Billing Amount` values also having Admission Type = 'Emergency').

### Implementation
See the code cell below.

```python
# Rows that have at least one missing value
rows_with_missing = df[df.isnull().any(axis=1)]
print("Rows with at least one missing value:", len(rows_with_missing))

# How many missing values does each affected row have?
print("\nDistribution of missing-value counts per row (for rows with at least 1 missing):")
print(df.isnull().sum(axis=1).value_counts().sort_index())

# Correlation between which columns tend to be missing together
print("\nCorrelation of missingness between columns:")
print(df.isnull().corr())
```

### Interpretation
If certain columns are consistently missing together (a high positive correlation near 1.0), that suggests a shared root cause — for example, a data-entry step that was skipped entirely for certain records — rather than independent random gaps.

# Part 2: Duplicate Records

## Identify Duplicates

### Concept
A duplicate record is a row that repeats the same information as another row, which can inflate counts and bias analysis if not addressed.

### Implementation
See the code cell below.

```python
duplicate_mask = df.duplicated()
print("Are there any exact duplicate rows?", duplicate_mask.any())
```

### Interpretation
This flags rows that are 100% identical, across every single column, to an earlier row.

## Count Duplicates

### Concept
Counting duplicates quantifies how significant the issue is relative to the size of the dataset.

### Implementation
See the code cell below.

```python
num_duplicates = df.duplicated().sum()
print("Number of Exact Duplicate Rows:", num_duplicates)
print("Percentage of Dataset:", round((num_duplicates / len(df)) * 100, 2), "%")
```

### Interpretation
Even a small percentage of duplicates can skew aggregate statistics like averages or category proportions if left unaddressed.

## Identify Completely Duplicated Rows

### Concept
Beyond counting, it is useful to actually view the duplicated rows to confirm they are true duplicates and not a data-loading artifact.

### Implementation
See the code cell below.

```python
fully_duplicated_rows = df[df.duplicated(keep=False)].sort_values(by=df.columns.tolist())
fully_duplicated_rows.head(10)
```

### Interpretation
Reviewing the actual rows (rather than just a count) helps confirm whether these are genuine repeated entries (e.g., the same admission logged twice) that should be dropped.

## Identify Potential Duplicate Records

### Concept
Records can be *near-duplicates* — not identical across every column, but matching on the fields that would realistically identify the same real-world event (e.g., same patient name and same admission date), while differing in a less critical field like billing amount rounding or a typo.

### Implementation
See the code cell below.

```python
subset_cols = ["Name", "Date of Admission", "Doctor"]
potential_duplicates = df[df.duplicated(subset=subset_cols, keep=False)].sort_values(by=subset_cols)
print("Potential duplicate records (same Name, Date of Admission, and Doctor):", len(potential_duplicates))
potential_duplicates.head(10)
```

### Interpretation
These near-duplicates are easy to miss with an exact-match check alone, but they represent the same likely real-world event and should be reviewed before being treated as independent records.

# Part 3: Invalid Values

## Negative Values Where Inappropriate

### Concept
Some columns can never legitimately be negative in the real world (e.g., age, billing amount, room number). Negative values in these columns indicate a data-entry or processing error.

### Implementation
See the code cell below.

```python
print("Negative Ages:", (df["Age"] < 0).sum())
print("Negative Billing Amounts:", (df["Billing Amount"] < 0).sum())
print("Negative Room Numbers:", (df["Room Number"] < 0).sum())
```

### Interpretation
Any non-zero count here points to values that must be corrected or removed, since a negative age or negative bill is not physically meaningful.

## Impossible Values

### Concept
An impossible value is one that falls outside any realistic real-world range, even if it is technically a valid positive number.

### Implementation
See the code cell below.

```python
print("Ages above 120:", (df["Age"] > 120).sum())
print("Ages equal to 0:", (df["Age"] == 0).sum())
print("Room Numbers equal to 0:", (df["Room Number"] == 0).sum())
print("Zero or unusually small Billing Amounts:", (df["Billing Amount"] <= 0).sum())
```

### Interpretation
Even without going negative, values like an age of 0 or 150 fall outside plausible human lifespans and should be flagged as likely data errors.

## Incorrect Categories

### Concept
Categorical columns should only contain values from a known, expected set. Any value outside that set is either a typo, inconsistent casing, or an invalid entry.

### Implementation
See the code cell below.

```python
expected_gender = {"Male", "Female"}
expected_admission_type = {"Emergency", "Elective", "Urgent"}
expected_test_results = {"Normal", "Abnormal", "Inconclusive"}

print("Unexpected Gender values:", set(df["Gender"].dropna().unique()) - expected_gender)
print("Unexpected Admission Type values:", set(df["Admission Type"].dropna().unique()) - expected_admission_type)
print("Unexpected Test Results values:", set(df["Test Results"].dropna().unique()) - expected_test_results)
```

### Interpretation
Any values printed here (e.g., 'male' in lowercase, or a stray typo) are inconsistent with the expected categories and should be standardized during cleaning.

## Invalid Dates

### Concept
Dates are invalid if they are malformed, in the future, or logically inconsistent with related date columns (e.g., a discharge date before the admission date).

### Implementation
See the code cell below.

```python
admission_dates = pd.to_datetime(df["Date of Admission"], errors="coerce")
discharge_dates = pd.to_datetime(df["Discharge Date"], errors="coerce")

print("Unparseable Admission Dates:", admission_dates.isnull().sum() - df["Date of Admission"].isnull().sum())
print("Unparseable Discharge Dates:", discharge_dates.isnull().sum() - df["Discharge Date"].isnull().sum())
print("Discharge Date before Admission Date:", (discharge_dates < admission_dates).sum())
print("Admission Dates in the future:", (admission_dates > pd.Timestamp.today()).sum())
```

### Interpretation
A discharge date earlier than the admission date is logically impossible and indicates either swapped columns or a data-entry mistake; unparseable or future dates point to formatting or entry errors.

## Incorrect Data Types

### Concept
A column's stored data type should match what it represents. Dates stored as plain text, or numbers stored as text, prevent proper sorting, filtering, and arithmetic.

### Implementation
See the code cell below.

```python
print("Current dtype of Date of Admission:", df["Date of Admission"].dtype)
print("Current dtype of Discharge Date:", df["Discharge Date"].dtype)
print("Current dtype of Age:", df["Age"].dtype)
print("Current dtype of Billing Amount:", df["Billing Amount"].dtype)
```

### Interpretation
`Date of Admission` and `Discharge Date` load as `object` (plain text) by default rather than an actual datetime type, meaning date-based operations (like calculating length of stay) will fail unless explicitly converted with `pd.to_datetime()`.

## Unexpected Values

### Concept
Unexpected values are extreme outliers that, while technically valid, sit far outside the normal range seen in the rest of the data and may warrant closer inspection.

### Implementation
See the code cell below.

```python
q1 = df["Billing Amount"].quantile(0.25)
q3 = df["Billing Amount"].quantile(0.75)
iqr = q3 - q1
lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

outliers = df[(df["Billing Amount"] < lower_bound) | (df["Billing Amount"] > upper_bound)]
print("Billing Amount outliers (IQR method):", len(outliers))
print("Valid range:", round(lower_bound, 2), "to", round(upper_bound, 2))
```

### Interpretation
Values falling outside the IQR-based bounds are statistical outliers; they aren't necessarily wrong, but should be reviewed to decide whether they're legitimate extreme cases or data errors.

## Empty Strings

### Concept
An empty string (`""`) is different from a true null value — it passes `isnull()` checks as "not missing" even though it carries no real information.

### Implementation
See the code cell below.

```python
text_columns = df.select_dtypes(include="object").columns
empty_string_counts = {col: (df[col] == "").sum() for col in text_columns}
empty_string_counts = {k: v for k, v in empty_string_counts.items() if v > 0}
print("Columns with empty strings:", empty_string_counts if empty_string_counts else "None found")
```

### Interpretation
Empty strings hide inside columns that `isnull()` would report as fully complete, so they need a separate, explicit check like this one.

## Null-like Values

### Concept
Null-like values are text placeholders (e.g., 'NA', 'N/A', 'null', 'None', 'unknown', '-') that represent missing data but are not recognized as null by pandas, since they are technically valid strings.

### Implementation
See the code cell below.

```python
null_like_tokens = ["na", "n/a", "null", "none", "unknown", "-", "missing", "nan"]

null_like_counts = {}
for col in text_columns:
    matches = df[col].astype(str).str.strip().str.lower().isin(null_like_tokens).sum()
    if matches > 0:
        null_like_counts[col] = matches

print("Columns with null-like placeholder text:", null_like_counts if null_like_counts else "None found")
```

### Interpretation
If any columns are flagged here, those values should be explicitly converted to real `NaN` values (e.g., with `df.replace(...)`) so they're correctly counted in the missing-value analysis from Part 1, rather than silently being treated as valid categories.

## Summary of Data Quality Findings

This notebook checked the Healthcare dataset for three categories of data-quality issues:

1. **Missing values** — counted and expressed as percentages per column, with an initial look at whether missingness is random or patterned.
2. **Duplicate records** — both exact full-row duplicates and near-duplicates based on `Name`, `Date of Admission`, and `Doctor`.
3. **Invalid values** — negative and impossible numeric values, non-standard categories, illogical or malformed dates, incorrect data types (dates stored as text), statistical outliers, empty strings, and null-like placeholder text.

These findings should guide the data-cleaning steps in the next stage of the project.
