# Notebook 2 – Data Inspection

This notebook performs an initial inspection of the dataset: its size, structure, data types, and a first look at its contents.

## Load the Dataset

```python
import pandas as pd

df = pd.read_csv("train.csv")
df.head()
```

## Number of Rows

### Concept
The number of rows tells us how many observations (records) are present in the dataset.

### Implementation
See the code cell below.

```python
num_rows = df.shape[0]
print("Number of Rows:", num_rows)
```

### Interpretation
This is the total count of passengers recorded in the dataset.

## Number of Columns

### Concept
The number of columns tells us how many features (variables) describe each observation.

### Implementation
See the code cell below.

```python
num_cols = df.shape[1]
print("Number of Columns:", num_cols)
```

### Interpretation
This is the total number of attributes recorded for each passenger.

## Data Types

### Concept
Each column has a data type (numeric, text/object, boolean, datetime, etc.) that determines what operations can be performed on it.

### Implementation
See the code cell below.

```python
print(df.dtypes)
```

### Interpretation
Columns with `int64`/`float64` are numerical, while columns with `object` are typically text or categorical.

## Column Names

### Concept
Knowing the exact column names is necessary before referencing them in code.

### Implementation
See the code cell below.

```python
print("Column Names:", df.columns.tolist())
```

### Interpretation
These are the exact labels used to access each feature in the dataset.

## Index

### Concept
The index labels each row. By default, pandas assigns a numeric range starting from 0.

### Implementation
See the code cell below.

```python
print(df.index)
```

### Interpretation
Unless explicitly set to an identifier column, the index is just a positional row label and carries no analytical meaning.

## head()

### Concept
`head()` displays the first few rows of the dataset, useful for a quick preview.

### Implementation
See the code cell below.

```python
df.head()
```

### Interpretation
This gives a quick sense of the columns present and the format of typical values.

## tail()

### Concept
`tail()` displays the last few rows of the dataset.

### Implementation
See the code cell below.

```python
df.tail()
```

### Interpretation
Checking the tail helps confirm the data loaded completely and looks consistent with the head.

## sample()

### Concept
`sample()` returns a random subset of rows, useful for spot-checking data outside the first/last rows.

### Implementation
See the code cell below.

```python
df.sample(5)
```

### Interpretation
A random sample helps verify that patterns seen in `head()`/`tail()` hold throughout the dataset.

## shape

### Concept
`shape` returns a tuple of (rows, columns), summarizing the dataset's size in one call.

### Implementation
See the code cell below.

```python
print("Shape:", df.shape)
```

### Interpretation
This confirms the dataset's dimensions in a single, convenient output.

## info()

### Concept
`info()` summarizes the dataframe: column names, non-null counts, and data types, along with memory usage.

### Implementation
See the code cell below.

```python
df.info()
```

### Interpretation
This is one of the fastest ways to spot missing values (columns with non-null counts less than the total row count) and confirm data types.

## describe()

### Concept
`describe()` provides summary statistics (count, mean, std, min, max, quartiles) for numerical columns by default.

### Implementation
See the code cell below.

```python
df.describe()
```

### Interpretation
This highlights the central tendency and spread of numerical features, and can reveal outliers (e.g., unusually high max values).

### describe() for categorical columns

### Concept
Passing `include="object"` summarizes categorical columns instead: count, unique values, the most frequent value, and its frequency.

### Implementation
See the code cell below.

```python
df.describe(include="object")
```

### Interpretation
This shows how many distinct categories exist in each text column and which category is most common.

## dtypes

### Concept
The `dtypes` attribute (as opposed to the `info()` method) returns a Series mapping each column to its data type, useful for programmatic filtering.

### Implementation
See the code cell below.

```python
print(df.dtypes)
```

### Interpretation
This can be used directly in code, e.g., to select all numeric columns with `df.select_dtypes(include="number")`.

## nunique()

### Concept
`nunique()` counts the number of unique values in each column, which helps distinguish categorical columns from identifiers or continuous numeric columns.

### Implementation
See the code cell below.

```python
print(df.nunique())
```

### Interpretation
Columns with very few unique values (e.g., `Sex`, `Embarked`, `Survived`) are strong candidates for categorical features. Columns where unique count equals row count (e.g., `PassengerId`) are identifiers.

## value_counts()

### Concept
`value_counts()` shows the frequency of each unique value in a column, useful for understanding class balance in categorical features.

### Implementation
See the code cell below.

```python
print(df["Sex"].value_counts())
print()
print(df["Embarked"].value_counts())
print()
print(df["Survived"].value_counts())
```

### Interpretation
This reveals the distribution of categories, e.g., whether the dataset is imbalanced (such as more non-survivors than survivors).

## Classifying Columns

Based on the inspection above, the columns can be grouped as follows.

### Numerical columns
- `Age`
- `Fare`
- `SibSp`
- `Parch`
- `Pclass` (technically numeric, but represents an ordinal category — passenger class 1st/2nd/3rd)

### Categorical columns
- `Sex`
- `Embarked`
- `Survived` (binary categorical target)
- `Pclass` (can also be treated as categorical/ordinal)

### Date/time columns
- None present in this dataset (no column stores a date or timestamp).

### Identifier columns
- `PassengerId` — a unique row identifier, not useful for analysis.
- `Ticket` — high-cardinality, mostly acts as a semi-identifier.
- `Name` — unique per passenger, acts as an identifier rather than a predictive feature (though titles could be extracted).

### Potential target columns
- `Survived` — the binary outcome (0 = did not survive, 1 = survived) is the natural prediction target for this dataset.
