## Load the Dataset

```python
import pandas as pd
import os

possible_names = ["healthcare_dataset.csv", "Healthcare_dataset.csv", "Healthcare Dataset.csv", "healthcare-dataset.csv"]
data_path = None

for name in possible_names:
    if os.path.exists(name):
        data_path = name
        break

if data_path is None:
    print("Could not find the dataset automatically.")
    print("Current folder:", os.getcwd())
    print("Files in this folder:", os.listdir())
    print("\nFix: set data_path manually below to the exact filename you see above,")
    print("e.g. data_path = 'my_actual_filename.csv'")
else:
    print("Found dataset at:", data_path)


df = pd.read_csv(data_path)
df.head()
```

## Number of Rows

### Concept
The number of rows tells us how many observations (records) are present in the dataset.

### Implementation

```python
num_rows = df.shape[0]
print("Number of Rows:", num_rows)
```

### Interpretation
This is the total count of patient admission records in the dataset.

## Number of Columns

### Concept
The number of columns tells us how many features (variables) describe each observation.

### Implementation

```python
num_cols = df.shape[1]
print("Number of Columns:", num_cols)
```

### Interpretation
This is the total number of attributes recorded for each patient admission.

## Data Types

### Concept
Each column has a data type (numeric, text/object, boolean, datetime, etc.) that determines what operations can be performed on it.

### Implementation
See the code cell below.

```python
print(df.dtypes)
```

### Interpretation
Columns with `int64`/`float64` are numerical, while columns with `object` are typically text or categorical. Note that the date columns may load as `object` (text) by default and would need explicit conversion with `pd.to_datetime()`.

Checking the tail helps confirm the data loaded completely and looks consistent with the head.

## sample()

### Concept
`sample()` returns a random subset of rows, useful for spot-checking data outside the first/last rows.

### Implementation

```python
df.sample(5)
```

A random sample helps verify that patterns seen in `head()`/`tail()` hold throughout the dataset, and can surface inconsistencies like mixed name casing.

## shape

### Concept
`shape` returns a tuple of (rows, columns), summarizing the dataset's size in one call.

### Implementation
See the code cell below.

```python
print("Shape:", df.shape)
```

## info()

### Concept
`info()` summarizes the dataframe: column names, non-null counts, and data types, along with memory usage.

### Implementation
See the code cell below.

```python
df.info()
```

### Interpretation
This is one of the fastest ways to spot missing values (columns with non-null counts less than the total row count) and confirm data types across all 15 columns.

## describe()

### Concept
`describe()` provides summary statistics (count, mean, std, min, max, quartiles) for numerical columns by default.

### Implementation
See the code cell below.

```python
df.describe()
```

### Interpretation
This highlights the central tendency and spread of `Age`, `Billing Amount`, and `Room Number`, and can reveal issues such as negative billing amounts or implausible ages.

### describe() for categorical columns

### Concept
Passing `include="object"` summarizes categorical columns instead: count, unique values, the most frequent value, and its frequency.

### Implementation
See the code cell below.

```python
df.describe(include="object")
```

### Interpretation
This shows how many distinct categories exist in columns like `Medical Condition` and `Insurance Provider`, and which category is most common.

## dtypes

### Concept
The `dtypes` attribute (as opposed to the `info()` method) returns a Series mapping each column to its data type, useful for programmatic filtering.

### Implementation
See the code cell below.

```python
print(df.dtypes)
```

### Interpretation
This can be used directly in code, e.g., to select all numeric columns with `df.select_dtypes(include="number")`.

## nunique()

### Concept
`nunique()` counts the number of unique values in each column, which helps distinguish categorical columns from identifiers or continuous numeric columns.

### Implementation
See the code cell below.

```python
print(df.nunique())
```

### Interpretation
Columns with very few unique values (e.g., `Gender`, `Admission Type`, `Test Results`) are strong candidates for categorical features. Columns where unique count is close to the row count (e.g., `Name`) are identifiers.

## value_counts()

### Concept
`value_counts()` shows the frequency of each unique value in a column, useful for understanding class balance in categorical features.

### Implementation
See the code cell below.

```python
print(df["Gender"].value_counts())
print()
print(df["Admission Type"].value_counts())
print()
print(df["Test Results"].value_counts())
```

### Interpretation
This reveals the distribution of categories, e.g., whether test results are roughly balanced across Normal, Abnormal, and Inconclusive, which matters for later modeling.
